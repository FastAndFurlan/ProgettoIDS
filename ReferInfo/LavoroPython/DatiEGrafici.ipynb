```python
import pandas as pd
data=pd.read_csv('Referendum_1946_2026.csv', sep=';')
anni=data.groupby(['ANNO_SPECIFICO', 'QUESITO']).agg({
    'ELETTORI':'sum',
    'VOTANTI':'sum',
    'NUMVOTISI':'sum',
    'NUMVOTINO':'sum'
}).reset_index()
nome_file_output = 'verifica_groupby_anni.csv'
anni.to_csv(nome_file_output, sep=';', index=False, encoding='utf-8-sig')
```
